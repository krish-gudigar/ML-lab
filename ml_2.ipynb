```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("ai_workforce_displacement_global_2020_2026.csv")
```

```python
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"]=(8,5)
```

# regression plot

```python
plt.style.use('dark_background')

plt.figure(facecolor='black')
plt.figure()
sns.regplot(data=df,
            x='record_id',
            y='net_workforce_change_pct',
            scatter_kws={'alpha':0.6,"color":"red"}
           )
plt.title("7.Regression Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df = pd.DataFrame({
    'record_id': range(1, 11),
    'net_workforce_change_pct': [2, 3, 5, 7, 6, 8, 7, 9, 10, 12]
})

plt.style.use('dark_background')

plt.figure(facecolor='black')

sns.regplot(
    data=df,
    x='record_id',
    y='net_workforce_change_pct',
    scatter_kws={'alpha': 0.6, "color": "red"},
    line_kws={"color": "cyan"} 
)

plt.title("7. Regression Plot", color='white')

plt.xlabel("Record ID", color='white')
plt.ylabel("Net Workforce Change (%)", color='white')

plt.show()
```

# 8. Count Plot

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df,
            y='income_group',
            order =df['income_group'].value_counts().index)
plt.title("8. Count Plot")
plt.show()
```

# Bar Plot

```python
plt.style.use('dark_background')

plt.figure(facecolor='black')
plt.figure()
sns.barplot(data=df,
            x='industry_sector',
            y='ai_adoption_index',
            estimator='mean'
           )
plt.xticks(rotation=45)
plt.title("9.Bar Plot")
plt.show()
```

# Pie chart

```python
plt.figure(figsize=(7,7))
df['industry_sector'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title("Pie chart")
```

```python
plt.figure(figsize=(10,5))
sns.violinplot(data=df,
               x='industry_sector',
               y='pct_sector_workforce_displaced')
plt.xticks(rotation=45)
plt.title("12.Violin Plot")
plt.show()
```
