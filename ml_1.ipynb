```python
num_cols=df.select_dtypes(include="number").columns
num_cols
```

# HISTOGRAM

```python
df[num_cols].hist(figsize=(16,12),bins=20)
plt.suptitle("1.Histogram of Numerical Features")
plt.tight_layout()
plt.show()
```

# Distribution Plot

```python
plt.figure()
sns.histplot(df['ai_adoption_index'],kde=True,color="steelblue")
plt.title('2.Distribution of AI Adoption Index')
plt.show()
```

# 3. Box Plot

```python
plt.figure(figsize=(15,6))
sns.boxplot(data=df[num_cols])
plt.xticks(rotation=90)
plt.title("3.Box Plot")
plt.show()
```

# correlation Heatmap

```python
plt.figure(figsize=(12,8))
sns.heatmap(df[num_cols].corr(),
            annot=True,
            cmap='coolwarm',
            fmt='.2f')
plt.title("4.Correlation Heatmap")
plt.show()
        
```

# Scatter Plot

```python
plt.figure()
sns.scatterplot(data=df,
                x='ai_adoption_index',
                y='pct_sector_workforce_displaced')
plt.title('5.AI Adoption vs Workforce Displacement')
plt.show()
```

# SCATTER PLOT WITH HUE

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=df,
                x='ai_adoption_index',
                y='pct_sector_workforce_displaced',
                hue='industry_sector')
plt.title('6.AI Adoption vs Workforce Displacement')
plt.show()
```
